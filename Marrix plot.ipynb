## Matrix plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
flight=sns.load_dataset('flights')

tips=sns.load_dataset('tips')
```

## Heat Map

```python
tips_co_relation=tips[['total_bill','tip','size']]
```

```python
tips_co_relation.corr()
```

```python
sns.heatmap(tips_co_relation.corr())
```

## Cluster Map


```python
sns.clustermap(tips_co_relation.corr())
```

## Pivot table 

```python
pvt_table=flight.pivot_table(values='passengers',index='month',columns='year')
```

```python
pvt_table
```

## Pivot table Heat Map

```python
sns.heatmap(pvt_table)
```
