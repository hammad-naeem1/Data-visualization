# saving a plot


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as pltimg
```

```python
x=np.array([[1,2,3],
            [4,5,6]])

y=np.array([[6,3,2],
            [5,8,1]])
```

## saving the plot

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.title('numbers plot')
plt.ylabel('y axis')
plt.xlabel('x axis')

# saving the plot
plt.savefig('saved plot')
```

## opening a plot

```python
img=pltimg.imread('saved plot.png')

plt.imshow(img)
plt.axis('off')
```
