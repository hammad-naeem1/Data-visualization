## subplot in matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.linspace(0,5,11)
y= x**2
```

#### creating subplots

```python
# after creating the subplot we need to give the plot command 
# just under it to plot something


plt.subplot(2,2,1)
plt.plot(x**2,y/2)


plt.subplot(2,2,4)
plt.plot(x,y)


plt.subplot(2,2,3)
plt.plot(y,x)

plt.subplot(2,2,2)
plt.plot(x**2,y**5)
```
