#####  object orianted way to subplot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.linspace(0,5,11)
y= x**2
```

### creating plot

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)

axis2=fig.add_axes([0.6,0.1,0.3,0.3])
axis2.plot(x,y)
```

### another one

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)

axis2=fig.add_axes([0.3,0.5,0.6,0.6])
axis2.plot(x,y)
```
