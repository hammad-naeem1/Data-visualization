## Distrubution plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### NOTE : we use distrubution plots to visualize the distrubution of quantative data

```python
data=sns.load_dataset('tips')

data['size'].unique()
```

#### DISTPLOT or HISTPLOT

```python
sns.histplot(data['total_bill'],kde=True)
plt.title('TOTAL_BILL')
```

## making subplot

```python
plt.subplot(1,3,1)
sns.histplot(data['total_bill'],kde=True)
plt.title('TOTAL_BILL')

plt.subplot(1,3,2)
sns.histplot(data['tip'],kde=True)
plt.title('Tips')


plt.subplot(1,3,3)
sns.histplot(data['size'],kde=True)
plt.title('SIZE')sns.rugplot(data)
```
