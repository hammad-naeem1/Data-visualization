## Distrubution plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### NOTE : we use distrubution plots to visualize the distrubution of quantative data

```python
data=sns.load_dataset('tips')

data['size'].unique()
```

#### DISTPLOT or HISTPLOT

```python
sns.histplot(data['total_bill'],kde=True)
plt.title('TOTAL_BILL')
```

## pair plot


```python
sns.pairplot(data)
```

```python
sns.pairplot(data,hue='size',palette='rainbow')
```

## joint plot


```python
sns.jointplot(x='total_bill',y='tip',data=data,kind='reg')                     
```

## Rug plot

```python
sns.rugplot(data)
```

```python
sns.rugplot(data['total_bill'])
```
