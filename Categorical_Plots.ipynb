## Categorical Data Plots


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=sns.load_dataset('tips')
```

```python
##                  Categorical Data Plots
#   There are a few main plot types for this:
#   boxplot  
#   violinplot
#   stripplot  
#   swarmplot   
#    barplot   
#    countplot   
```

## BOX PLOT

```python
sns.boxplot(x=data['tip'],y=data['sex'],data=data)
```

```python
sns.boxplot(x=data['tip'],y=data['sex'],data=data,palette='rainbow')
```

## count plot

```python
sns.countplot(x=data['sex'],hue=data['smoker'])
```

## Bar plot

```python
sns.barplot(y=data['total_bill'],x=data['smoker'])
```

```python
import numpy as np
sns.barplot(y=data['tip'],x=data['sex'],estimator=np.sum)
```

##  violinplot

```python
sns.violinplot(x=data['day'],y=data['total_bill'])
```

## stripplot

```python
sns.stripplot(x=data['day'],y=data['total_bill'])
```

## Swarmplot

```python
sns.swarmplot(x=data['day'],y=data['total_bill'])
```
